# Supply Chain Basics

This notebook introduces supply chains with simple examples.

## What is a Supply Chain?

A supply chain is the journey of a product from where it is made to where it is bought.

Let us trace a chocolate bar:

```python
# Let us trace the supply chain of a chocolate bar

chocolate_supply_chain = [
    '1. Farmers grow cocoa beans',
    '2. Trucks take beans to factory',
    '3. Factory makes chocolate',
    '4. Chocolate is packed in boxes',
    '5. Trucks take boxes to store',
    '6. You buy the chocolate bar',
]

for step in chocolate_supply_chain:
    print(step)
```

## The 5 Steps

Every supply chain has these steps: Plan, Source, Make, Deliver, Return.

```python
# The 5 steps of a supply chain
steps = {
    'Plan': 'Decide what to make and how much',
    'Source': 'Find materials and suppliers',
    'Make': 'Turn materials into a product',
    'Deliver': 'Move the product to customers',
    'Return': 'Handle returns and recycling',
}

for step, description in steps.items():
    print(f'{step}: {description}')
```

## Example: Pizza Shop

Let us trace a pizza supply chain:

```python
# Pizza supply chain
pizza_chain = {
    'plan': 'Decide how many pizzas to make today',
    'source': 'Buy flour, cheese, tomatoes, pepperoni',
    'make': 'Cook the pizza in the oven',
    'deliver': 'Deliver to customer house',
    'return': 'Take back empty box for recycling',
}

for step, action in pizza_chain.items():
    print(f'{step}: {action}')
```
