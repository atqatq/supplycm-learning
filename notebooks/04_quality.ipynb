## What is Six Sigma?

Six Sigma is a quality goal: only 3.4 defects per million products.

That is 99.99966% perfect.

````python
# How good is Six Sigma?
six_sigma_dpmo = 3.4
six_sigma_yield = 1 - (six_sigma_dpmo / 1_000_000)
print(f'Six Sigma DPMO: {six_sigma_dpmo}')
print(f'Six Sigma yield: {six_sigma_yield * 100:.5f}%')
print(f'That means only {six_sigma_dpmo} bad products per million!')
````

## Measuring Quality: DPMO

```python
dpmo = (defects / (units * opportunities)) * 1_000_000
```

````python
# Toy factory example
units = 1000           # toys made
opportunities = 10     # ways each toy can be defective
defects = 50           # defects found

dpmo_value = dpmo(defects, units, opportunities)
sigma = sigma_level(dpmo_value)

print(f'Units made: {units}')
print(f'Opportunities per unit: {opportunities}')
print(f'Defects found: {defects}')
print(f'DPMO: {dpmo_value:.0f}')
print(f'Sigma level: {sigma:.1f}')
print(f'\nSix Sigma target: 6.0')
print(f'Your level: {sigma:.1f} ({"Good!" if sigma >= 4 else "Needs improvement"})')
````

## Process Capability

Can your process meet customer specifications?

````python
# Customer wants toys weighing between 8 and 12 grams
upper_spec = 12     # max acceptable weight
lower_spec = 8      # min acceptable weight
mean = 10           # your process average
std_dev = 0.5       # your process variation

cp = process_capability_cp(upper_spec, lower_spec, std_dev)
cpk = process_capability_cpk(upper_spec, lower_spec, mean, std_dev)

print(f'Specification: {lower_spec} to {upper_spec} grams')
print(f'Process mean: {mean} grams')
print(f'Std dev: {std_dev} grams')
print(f'Cp: {cp:.2f} (potential)')
print(f'Cpk: {cpk:.2f} (actual)')
print()
if cpk >= 1.33:
    print('Process is CAPABLE')
elif cpk >= 1.0:
    print('Process is MARGINAL')
else:
    print('Process is NOT CAPABLE')
````

## Control Charts

Watch your process over time.

````python
# Daily samples of toy weights
samples = [
    [10.1, 10.2, 9.9],   # day 1
    [10.0, 10.1, 10.2],   # day 2
    [10.3, 10.1, 10.0],   # day 3
    [9.8, 9.9, 10.0],     # day 4
    [10.2, 10.3, 10.1],   # day 5
]

means, upper_limit, lower_limit = x_bar_chart(samples)

print('Day  Average  Status')
print('-' * 30)
for i, (mean_val) in enumerate(means):
    if mean_val > upper_limit or mean_val < lower_limit:
        status = 'OUT OF CONTROL!'
    else:
        status = 'OK'
    print(f'{i+1:>3}  {mean_val:.2f}    {status}')

print(f'\nUpper limit: {upper_limit:.2f}')
print(f'Lower limit: {lower_limit:.2f}')
````
