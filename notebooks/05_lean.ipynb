## The 8 Wastes (DOWNTIME)

Remember with DOWNTIME:

````python
wastes = {
    'D': 'Defects - products that are wrong',
    'O': 'Overproduction - making more than needed',
    'W': 'Waiting - time spent idle',
    'N': 'Non-utilized talent - not using skills',
    'T': 'Transportation - moving stuff unnecessarily',
    'I': 'Inventory - keeping too much stock',
    'M': 'Motion - people moving too much',
    'E': 'Excess processing - doing more than needed',
}

for letter, waste in wastes.items():
    print(f'{letter}: {waste}')
````

## Takt Time

The pace of customer demand.

```python
takt_time = available_time / customer_demand
```

````python
# Coffee shop example
available_time = 480   # 8 hours = 480 minutes
customer_demand = 60   # 60 customers per day

takt = takt_time(available_time, customer_demand)
print(f'Available time: {available_time} minutes')
print(f'Customer demand: {customer_demand} cups')
print(f'Takt time: {takt:.1f} minutes per cup')
print(f'\nYou need to make one cup every {takt:.1f} minutes')
````

## OEE: Overall Equipment Effectiveness

```python
oee = availability * performance * quality
```

World class OEE is 85%.

````python
# Espresso machine efficiency
availability = 0.85   # running 85% of time
performance = 0.90   # 90% of max speed
quality = 0.95        # 95% good cups

oee_value = oee(availability, performance, quality)
print(f'Availability: {availability*100:.0f}%')
print(f'Performance: {performance*100:.0f}%')
print(f'Quality: {quality*100:.0f}%')
print(f'OEE: {oee_value*100:.1f}%')
print()

if oee_value >= 0.85:
    print('World class!')
elif oee_value >= 0.60:
    print('Typical')
else:
    print('Room for improvement')
````

## Little's Law

```python
work_in_progress = throughput_rate * flow_time
```

````python
# Bakery example
throughput_rate = 10   # 10 loaves per hour
flow_time = 4           # each loaf takes 4 hours

wip = wip_calculation(throughput_rate, flow_time)
print(f'Throughput: {throughput_rate} loaves/hour')
print(f'Flow time: {flow_time} hours')
print(f'WIP (bread in progress): {wip} loaves')
````

## Cycle Time Efficiency

```python
efficiency = value_added_time / total_cycle_time
```

Most processes are less than 10% efficient!

````python
# How efficient is the bakery?
value_added_time = 0.5   # 30 minutes of actual work
total_cycle_time = 4     # 4 hours total

efficiency = cycle_time_efficiency(value_added_time, total_cycle_time)
print(f'Value-added time: {value_added_time} hours')
print(f'Total cycle time: {total_cycle_time} hours')
print(f'Efficiency: {efficiency*100:.0f}%')
print(f'\n{(1-efficiency)*100:.0f}% of time is wasted!')
````
