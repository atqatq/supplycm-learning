# Complete Supply Chain Example

Put everything together in a real business scenario.

## The Business: A Coffee Shop

You run a coffee shop. Let us plan the entire supply chain.

```python
# Import everything we need
from supplycm.forecasting import single_exponential_smoothing
from supplycm.inventory import economic_order_quantity, safety_stock_normal, reorder_point
from supplycm.supplier import topsis
from supplycm.quality import dpmo, sigma_level
from supplycm.lean import takt_time, oee
from supplycm.sop import production_level_strategy
from supplycm.contracts import revenue_sharing_contract
from supplycm.sustainability import carbon_footprint_transport

print('All modules loaded!')
```

## Step 1: Forecast Demand

```python
# Last 8 weeks of daily cup sales
weekly_sales = [350, 360, 340, 370, 380, 375, 390, 385]
forecast = single_exponential_smoothing(weekly_sales, alpha=0.3)
print(f'Weekly sales: {weekly_sales}')
print(f'Next week forecast: {forecast[-1]:.0f} cups')
print(f'Annual demand: {forecast[-1] * 50:.0f} cups (50 weeks)')
```

## Step 2: Inventory (Coffee Beans)

Each cup uses 0.05 lbs of beans.

```python
annual_cups = forecast[-1] * 50
annual_beans = annual_cups * 0.05  # pounds

eoq = economic_order_quantity(annual_beans, 25, 2)
ss = safety_stock_normal(1.96, 5, 1)
weekly_demand = annual_beans / 50
rop = reorder_point(weekly_demand, 1, ss)

print(f'Annual bean demand: {annual_beans:.0f} pounds')
print(f'EOQ: {eoq:.0f} pounds per order')
print(f'Safety stock: {ss:.0f} pounds')
print(f'Reorder point: {rop:.0f} pounds')
```

## Step 3: Choose Supplier

```python
# 3 bean suppliers
ranking = topsis(
    [[80, 90, 85], [70, 85, 90], [85, 80, 70]],
    [0.4, 0.35, 0.25],
    ['benefit', 'benefit', 'benefit']
)
suppliers = ['Supplier A', 'Supplier B', 'Supplier C']
print(f'Best supplier: {suppliers[ranking[0]]}')
```

## Step 4: Quality

```python
# 5000 cups, 5 checks each, 20 defects
dpmo_value = dpmo(20, 5000, 5)
sigma = sigma_level(dpmo_value)
print(f'DPMO: {dpmo_value:.0f}')
print(f'Sigma level: {sigma:.1f}')
```

## Step 5: Lean

```python
# Open 8 hours = 480 min, 60 cups/day
takt = takt_time(480, 60)
oee_value = oee(0.90, 0.95, 0.98)
print(f'Takt time: {takt:.1f} min/cup')
print(f'OEE: {oee_value*100:.1f}%')
```

## Step 6: Contract

```python
result = revenue_sharing_contract(7, 4, 0.10, annual_cups, 5)
print(f'Supplier profit: ${result["supplier_profit"]:,.0f}')
print(f'Your profit: ${result["retailer_profit"]:,.0f}')
```

## Step 7: Sustainability

```python
# Beans travel 500 km by truck, 0.34 tonnes per year
co2 = carbon_footprint_transport(500, 0.34)
print(f'Annual delivery CO2: {co2:.1f} kg')
```
