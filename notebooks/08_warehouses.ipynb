## ABC Slotting

Put popular items near the door.

```python
# 6 products with monthly sales
items = [
    ('Popular Toy', 1000, 1),  # name, demand, size
    ('Medium Toy', 500, 1),
    ('Slow Toy', 100, 1),
    ('Big Toy', 50, 2),
    ('Small Toy', 200, 1),
    ('Tiny Toy', 20, 1),
]

slotting = warehouse_slotting_abc(items, num_zones=3)
print('Warehouse slotting (zone 0 = closest to door):')
for item, zone in slotting:
    print(f'  {item}: Zone {zone}')
```

## Order Picking Route

Plan the shortest route to pick items.

```python
# Locations of items to pick (x, y coordinates)
pick_locations = [
    (1, 1),   # item 1
    (3, 5),   # item 2
    (2, 8),   # item 3
    (7, 2),   # item 4
    (5, 6),   # item 5
]
depot = (0, 0)  # start and end here

route = traveling_salesman_picking(pick_locations, depot)
print('Pick items in this order:')
for i, step in enumerate(route):
    if step < len(pick_locations):
        loc = pick_locations[step]
        print(f'  Step {i+1}: Go to {loc} (item {step+1})')
    else:
        print(f'  Step {i+1}: Return to depot')
```

## Pallet Building

Stack boxes on pallets efficiently.

```python
# Items as (length, width, height)
items = [
    (2, 2, 2),  # big box
    (1, 1, 1),  # small box
    (3, 3, 3),  # very big box
    (1, 1, 1),  # small box
    (2, 2, 2),  # big box
]
pallet_capacity = 8  # cubic units

pallets = pallet_building(items, pallet_capacity)
print(f'Number of pallets needed: {len(pallets)}')
for i, pallet in enumerate(pallets):
    print(f'  Pallet {i+1}: {len(pallet)} items')
```
