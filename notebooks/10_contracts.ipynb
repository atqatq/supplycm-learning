# Supply Chain Contracts

Learn how contracts align buyer and supplier incentives.

```python
from supplycm.contracts import revenue_sharing_contract, buyback_contract, quantity_flexibility_contract
print('Ready!')
```

## Revenue Sharing Contract

Supplier charges lower price but gets a share of sales.

```python
# Supplier sells to retailer for $30 (instead of $40)
# Retailer sells to customer for $70
# Retailer shares 20% of revenue with supplier
result = revenue_sharing_contract(
    wholesale_price=30,
    retail_price=70,
    revenue_share_fraction=0.20,
    demand=1000,
    unit_cost=20
)
print('Revenue Sharing Contract:')
print(f'  Supplier profit: ${result["supplier_profit"]:,.0f}')
print(f'  Retailer profit: ${result["retailer_profit"]:,.0f}')
print(f'  Total profit:    ${result["total_profit"]:,.0f}')
```

## Buyback Contract

Supplier buys back unsold inventory.

```python
# Retailer orders 1000 units but only sells 800
# Supplier buys back 200 unsold at $25 each
result = buyback_contract(
    wholesale_price=40,
    retail_price=80,
    buyback_price=25,
    demand=800,
    unit_cost=20,
    salvage_value=10
)
print('Buyback Contract:')
print(f'  Supplier profit: ${result["supplier_profit"]:,.0f}')
print(f'  Retailer profit: ${result["retailer_profit"]:,.0f}')
print(f'  Total profit:    ${result["total_profit"]:,.0f}')
```

## Quantity Flexibility Contract

Retailer can adjust order within a range.

```python
# Retailer ordered 1000, demand was 900
# Contract allows 20% adjustment
result = quantity_flexibility_contract(
    wholesale_price=40,
    retail_price=80,
    order_quantity=1000,
    actual_demand=900,
    unit_cost=20,
    flexibility_fraction=0.20
)
print('Quantity Flexibility Contract:')
print(f'  Final quantity: {result["final_quantity"]}')
print(f'  Supplier profit: ${result["supplier_profit"]:,.0f}')
print(f'  Retailer profit: ${result["retailer_profit"]:,.0f}')
```

## Compare All Three

Which contract is best depends on the situation.

```python
# Compare contracts for the same scenario
print('Contract Comparison:')
print(f'{"Contract":<25} {"Supplier":>10} {"Retailer":>10} {"Total":>10}')
print('-' * 58)

# Revenue sharing
rs = revenue_sharing_contract(30, 70, 0.20, 1000, 20)
print(f'{"Revenue Sharing":<25} ${rs["supplier_profit"]:>8,.0f} ${rs["retailer_profit"]:>8,.0f} ${rs["total_profit"]:>8,.0f}')

# Buyback
bb = buyback_contract(40, 80, 25, 800, 20, 10)
print(f'{"Buyback":<25} ${bb["supplier_profit"]:>8,.0f} ${bb["retailer_profit"]:>8,.0f} ${bb["total_profit"]:>8,.0f}')

print('\nChoose the one with highest total profit!')
```
