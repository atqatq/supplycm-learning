## Carbon Footprint

```python
co2_emissions = distance * weight * emission_factor
```

````python
# Truck delivers 10 tonnes over 500 km
co2 = carbon_footprint_transport(distance=500, weight=10)
print(f'Distance: 500 km')
print(f'Weight: 10 tonnes')
print(f'CO2 emissions: {co2:.1f} kg')
````

## Compare Transport Modes

````python
# Same delivery, different modes
modes = {
    'Truck': 0.062,
    'Train': 0.022,
    'Ship': 0.016,
    'Air': 0.602,
}

print('CO2 by transport mode (500 km, 10 tonnes):')
for mode, factor in modes.items():
    co2 = carbon_footprint_transport(500, 10, factor)
    print(f'  {mode}: {co2:.0f} kg CO2')
````

## Reverse Logistics

Handling returns and recycling.

````python
# 10% of products returned
result = reverse_logistics_cost(
    return_rate=0.10,
    unit_cost=100,
    processing_cost=5,
    disposal_cost=2,
    resale_value=30
)
print(f'Return rate: 10%')
print(f'Net cost per unit sold: ${result:.2f}')
````
