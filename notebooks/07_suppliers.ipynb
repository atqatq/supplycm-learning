## Choosing Criteria Weights with AHP

Compare criteria in pairs to get weights.

```python
# Compare 3 criteria: Price, Quality, Delivery
# 1 = equal, 3 = slightly more, 5 = much more important
pairwise = [
    [1,     3,   5],   # Price vs Quality, Price vs Delivery
    [1/3,   1,   3],   # Quality vs Price, Quality vs Delivery
    [1/5, 1/3,   1],   # Delivery vs Price, Delivery vs Quality
]

weights = ahp_supplier_selection(pairwise)
criteria = ['Price', 'Quality', 'Delivery']

print('Criteria weights:')
for c, w in zip(criteria, weights):
    print(f'  {c}: {w:.2f} ({w*100:.0f}%)')
print(f'  Total: {sum(weights):.2f}')
```

## Ranking Suppliers with TOPSIS

```python
# 3 suppliers scored on each criterion (0-100)
decision_matrix = [
    [80, 90, 85],  # Supplier A
    [70, 85, 90],  # Supplier B
    [85, 80, 70],  # Supplier C
]

ranking = topsis(decision_matrix, weights, ['benefit', 'benefit', 'benefit'])

suppliers = ['Supplier A', 'Supplier B', 'Supplier C']
print('Supplier ranking (best to worst):')
for rank, idx in enumerate(ranking, 1):
    scores = decision_matrix[idx]
    print(f'  {rank}. {suppliers[idx]} - scores: {scores}')
```

## Supplier Segmentation (Kraljic Matrix)

Sort suppliers by risk and profit impact.

```python
suppliers_data = [
    ('Engine Maker', 0.9, 0.8),   # high profit, high risk
    ('Box Supplier', 0.7, 0.2),   # high profit, low risk
    ('Rare Parts Co', 0.2, 0.9),  # low profit, high risk
    ('Office Store', 0.3, 0.3),   # low profit, low risk
]

result = supplier_segmentation(suppliers_data)
print('Supplier classification:')
for name, group in result:
    print(f'  {name}: {group}')
```

## Measuring On-Time Delivery

```python
# Track delivery performance
deliveries = [
    (10, 9),   # due day 10, arrived day 9 (early)
    (15, 15),  # due day 15, arrived day 15 (on time)
    (20, 22),  # due day 20, arrived day 22 (LATE)
    (25, 24),  # due day 25, arrived day 24 (early)
]

rate = on_time_delivery_rate(deliveries)
print('Delivery performance:')
for due, actual in deliveries:
    status = 'ON TIME' if actual <= due else 'LATE'
    print(f'  Due: {due}, Arrived: {actual} - {status}')
print(f'\nOn-time delivery rate: {rate*100:.1f}%')
```
