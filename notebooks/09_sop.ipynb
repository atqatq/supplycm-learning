## Demand vs Supply

Compare what customers want with what you can make.

```python
# 5 months of demand and supply
demand = [100, 120, 130, 110, 125]
supply = [110, 110, 130, 130, 130]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']

net, cumulative = demand_supply_match(demand, supply)
print('Month  Demand  Supply  Gap  Cumulative')
print('-' * 45)
for i, month in enumerate(months):
    gap = net[i]
    status = 'shortage' if gap > 0 else 'surplus'
    print(f'{month:>5}  {demand[i]:>6}  {supply[i]:>6}  {gap:>4}  {cumulative[i]:>8} ({status})')
```

## Chase Strategy

Production matches demand each period.

```python
chase_plan = production_chase_strategy(demand)
print('Chase Strategy (production matches demand):')
for i, month in enumerate(months):
    print(f'  {month}: produce {chase_plan[i]:.0f}')
print('\nPros: Low inventory')
print('Cons: Workforce fluctuates')
```

## Level Strategy

Constant production rate.

```python
level_plan = production_level_strategy(demand)
print('Level Strategy (constant production):')
for i, month in enumerate(months):
    diff = level_plan[i] - demand[i]
    status = 'surplus' if diff > 0 else 'shortage'
    print(f'  {month}: produce {level_plan[i]:.0f}, demand {demand[i]}, {status}')
print('\nPros: Stable workforce')
print('Cons: Higher inventory')
```

## Which to Choose?

| Product Type | Best Strategy | Reason |
|-------------|---------------|--------|
| Perishable (food) | Chase | Cannot store extra |
| Non-perishable | Level | Can store extra |
| Seasonal | Chase | Demand changes a lot |
| Stable demand | Level | Easy to plan |
