# Forecasting Demand

Learn how to predict future sales using simple methods.

## Setup

First, install and import supplycm:

```python
# Install supplycm if you have not already
# Run this in your terminal: pip install supplycm

from supplycm.forecasting import (
    naive_forecast,
    simple_moving_average,
    single_exponential_smoothing,
)
from supplycm.statistics import mape

print('Ready!')
```

## Your Sales Data

Imagine you sell ice cream. Here are your sales for 7 days:

```python
# Daily ice cream sales
sales = [40, 45, 42, 50, 48, 55, 52]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

print('Day  Sales')
print('-' * 12)
for day, sale in zip(days, sales):
    print(f'{day}:  {sale} cones')
```

## Method 1: Naive Forecast

The simplest method. Tomorrow equals today.

```python
# Naive forecast: each day prediction = previous day actual
naive = naive_forecast(sales)
print('Naive forecast:')
for i, (day, actual, forecast) in enumerate(zip(days, sales, naive)):
    f = f'{forecast:.0f}' if forecast else 'N/A'
    print(f'  {day}: actual={actual}, forecast={f}')

print(f'\nForecast for next Monday: {naive[-1]} cones')
```

## Method 2: Moving Average

Average the last few days. Smooths out ups and downs.

```python
# 3-day moving average
window = 3
sma = simple_moving_average(sales, window=window)
print(f'{window}-day Moving Average:')
for i, (day, actual, forecast) in enumerate(zip(days, sales, sma)):
    f = f'{forecast:.1f}' if forecast else 'N/A'
    print(f'  {day}: actual={actual}, forecast={f}')

print(f'\nForecast for next Monday: {sma[-1]:.1f} cones')
```

## Method 3: Exponential Smoothing

Recent days matter more. Alpha controls how much.

- alpha = 0.1: smooth (old data matters a lot)
- alpha = 0.5: balanced
- alpha = 0.9: jumpy (recent data matters most)

```python
# Try different alpha values
for alpha in [0.1, 0.3, 0.5]:
    ses = single_exponential_smoothing(sales, alpha=alpha)
    print(f'alpha={alpha}: forecast for Monday = {ses[-1]:.1f} cones')
```

## Which Method is Best?

Use MAPE to check accuracy. Lower is better.

```python
# Compare methods using last 3 days as test
train = sales[:-3]  # first 4 days
test = sales[-3:]    # last 3 days

# Naive forecast for test days
naive_test = [train[-1]] * 3
# Moving average for test days
sma_train = simple_moving_average(train, window=3)
sma_test = [sma_train[-1]] * 3
# Exponential smoothing for test days
ses_train = single_exponential_smoothing(train, alpha=0.3)
ses_test = [ses_train[-1]] * 3

print(f'Actual sales:      {test}')
print(f'Naive forecast:    {naive_test}')
print(f'Moving avg forecast: {sma_test}')
print(f'Exp smoothing:     {[round(f,1) for f in ses_test]}')
print()
print(f'MAPE Naive:      {mape(test, naive_test):.1f}%')
print(f'MAPE Moving Avg: {mape(test, sma_test):.1f}%')
print(f'MAPE Exp Smooth: {mape(test, ses_test):.1f}%')

print('\nLowest MAPE = best method')
```
