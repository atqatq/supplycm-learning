# Inventory Management

Learn how much to order and when to reorder.

## Setup

````python
from supplycm.inventory import (
    economic_order_quantity,
    safety_stock_normal,
    reorder_point,
    abc_analysis,
)
print('Ready!')
````

## Step 1: Economic Order Quantity (EOQ)

EOQ tells you the best amount to order each time.

```python
eoq = (2 * annual_demand * ordering_cost / holding_cost) ** 0.5
```

````python
# Toy store example
annual_demand = 12000   # toys sold per year
ordering_cost = 100     # dollars per order
holding_cost = 5        # dollars per toy per year

eoq = economic_order_quantity(annual_demand, ordering_cost, holding_cost)
print(f'Annual demand: {annual_demand} toys')
print(f'Ordering cost: ${ordering_cost}')
print(f'Holding cost: ${holding_cost}/toy/year')
print(f'EOQ: {eoq:.0f} toys per order')
print(f'Orders per year: {annual_demand / eoq:.1f}')
print(f'Days between orders: {365 / (annual_demand / eoq):.0f}')
````

## Step 2: Safety Stock

Extra inventory to protect against surprises.

```python
import math
safety_stock = z_score * demand_std * (lead_time ** 0.5)
```

````python
# Safety stock for 97.5% service level
z_score = 1.96        # 97.5% service level
demand_std = 30       # demand varies by 30 per week
lead_time = 2         # supplier takes 2 weeks

safety = safety_stock_normal(z_score, demand_std, lead_time)
print(f'Z-score: {z_score} (97.5% sure)')
print(f'Demand std: {demand_std} units/week')
print(f'Lead time: {lead_time} weeks')
print(f'Safety stock: {safety:.0f} units')
````

## Step 3: Reorder Point

When inventory drops to this number, order more.

```python
reorder_point = (demand_per_week * lead_time) + safety_stock
```

````python
# Reorder point
demand_per_week = annual_demand / 52  # weekly demand
rop = reorder_point(demand_per_week, lead_time, safety)

print(f'Weekly demand: {demand_per_week:.0f} toys')
print(f'Lead time: {lead_time} weeks')
print(f'Safety stock: {safety:.0f} toys')
print(f'Reorder point: {rop:.0f} toys')
print(f'\nWhen inventory drops to {rop:.0f}, order {eoq:.0f} more toys')
````

## Step 4: ABC Analysis

Not all products are equal. Sort them by value.

````python
# ABC analysis of 6 products
products = [
    ('Toy A', 50000),
    ('Toy B', 30000),
    ('Toy C', 10000),
    ('Toy D', 5000),
    ('Toy E', 3000),
    ('Toy F', 2000),
]

result = abc_analysis(products)
print(f'{"Product":<10} {"Value":>10} {"Class":>6}')
print('-' * 30)
for product, group, cumulative in result:
    value = dict(products)[product]
    print(f'{product:<10} ${value:>9,} {group:>6}')

print('\nClass A = top sellers (watch closely)')
print('Class C = slow sellers (do not spend much time)')
````
