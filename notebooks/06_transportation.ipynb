Learn how to plan delivery routes.

## Traveling Salesman Problem (TSP)

Find the shortest route visiting all locations and returning to start.

```python
# 5 delivery locations
distances = [
    [0,  10, 15, 20, 25],  # from location 0
    [10, 0,  35, 25, 30],  # from location 1
    [15, 35, 0,  30, 20],  # from location 2
    [20, 25, 30, 0,  15],  # from location 3
    [25, 30, 20, 15, 0],   # from location 4
]

locations = ['Warehouse', 'House A', 'House B', 'House C', 'House D']

route, total = tsp_nearest_neighbor(distances)
print('Nearest Neighbor Route:')
for i in route:
    print(f'  -> {locations[i]}')
print(f'Total distance: {total} km')
```

## Improve with 2-opt

```python
route2, total2 = tsp_two_opt(distances)
print('2-opt Improved Route:')
for i in route2:
    print(f'  -> {locations[i]}')
print(f'Total distance: {total2} km')
print(f'Improvement: {((total - total2) / total * 100):.1f}%')
```

## Vehicle Routing Problem (VRP)

Multiple trucks with capacity limits.

```python
# 5 customers need delivery
demands = [0, 5, 10, 8, 12]  # customer 0 is warehouse
vehicle_capacity = 15

routes = vrp_capacitated_greedy(distances, demands, vehicle_capacity)
print(f'Vehicle capacity: {vehicle_capacity} units')
print(f'Number of routes: {len(routes)}')
for i, route in enumerate(routes):
    route_demand = sum(demands[j] for j in route)
    route_names = [locations[j] for j in route]
    print(f'  Truck {i+1}: {" -> ".join(route_names)} (demand: {route_demand})')
```

## Assignment Problem

Assign workers to jobs at minimum cost.

```python
# 3 workers, 3 jobs, cost matrix
costs = [
    [10, 15, 20],  # Worker 0 costs for jobs 0, 1, 2
    [5,  12, 8],   # Worker 1
    [14, 7,  11],  # Worker 2
]

assignments, total_cost = assignment_problem_hungarian(costs)
print('Optimal assignments:')
for worker, job in assignments:
    print(f'  Worker {worker} -> Job {job} (cost: {costs[worker][job]})')
print(f'Total cost: {total_cost}')
```
